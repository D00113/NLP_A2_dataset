```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_files = [
    "handwriting.png",
    "cat.jpg",
    "brickwall.jpg",
    "parrot.jpg",
    "girl.png"
]

images = {}

for file in image_files:
    if not os.path.exists(file):
        print(f"File not found: {file}")
        continue

    img = cv2.imread(file)
    if img is None:
        print(f"Failed to load image: {file}")
        continue

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    images[file] = img_rgb

plt.figure(figsize=(15, 8))
for idx, (filename, img) in enumerate(images.items()):
    plt.subplot(2, 3, idx + 1)
    plt.imshow(img)
    plt.title(filename)
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
processed_images = {}

for name, img in images.items():
    if len(img.shape) == 3:
        gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray_img = img

    resized_img = cv2.resize(gray_img, (256, 256))

    processed_images[name] = resized_img

    plt.figure()
    plt.title(f'Processed: {name}')
    plt.imshow(resized_img, cmap='gray')
    plt.axis('off')
    plt.show()
```
